```python

# Liste des URLs (extraites de vos commandes curl)


import os
import requests

def download_files(url_list, rndta: str):
    # 1. Création du dossier s'il n'existe pas
    if not os.path.exists(rndta):
        os.makedirs(rndta)
        print(f"📁 Dossier '{rndta}' créé.")
    else:
        print(f"📂 Utilisation du dossier existant : '{rndta}'")

    ##### Clean repo
    """Supprime les fichiers de données existants dans le dossier cible avant le téléchargement."""
    ext = ('.ctl', '.bin', '.dat')
    print(f"🧹 Cleaning repository in '{rndta}'...")
    
    # On itère maintenant sur le contenu du dossier cible
    for filename in os.listdir(rndta):
        if filename.lower().endswith(ext):
            file_path = os.path.join(rndta, filename)
            try:
                os.remove(file_path)
                print(f"   Supprimé : {filename}")
            except Exception as e:
                print(f"   ❌ Erreur suppression {filename}: {e}")

    #### Download phase
    for url in url_list:
        filename = url.split('/')[-1]
        # On définit le chemin complet (dossier + nom de fichier)
        save_path = os.path.join(rndta, filename)
        
        print(f"⏳ Téléchargement de {filename} vers {rndta}...")
        
        try:
            with requests.get(url, stream=True) as r:
                r.raise_for_status() 
                
                with open(save_path, 'wb') as f:
                    for chunk in r.iter_content(chunk_size=8192):
                        if chunk:
                            f.write(chunk)
                            
            print(f"✅ Terminé : {filename}")
            
        except requests.exceptions.RequestException as e:
            print(f"❌ Erreur lors du téléchargement de {filename} : {e}")

if __name__ == "__main__":
    download_files(urls_spi_rfe2, "rfe2")
```

```python


# Configuration
# Mettez le chemin vers le dossier où sont stockés vos fichiers .ctl
directory = "./" 
new_path = "//wsl$/Ubuntu/home/elmamoune/cmt/CMTK_senegal/ClimToolKit/data/runoff/"

def update_all_ctls(folder, absolute_path):
    # On récupère TOUS les fichiers .ctl sans exception
    files = [f for f in os.listdir(folder) if f.endswith('.ctl')]
    
    if not files:
        print("Aucun fichier .ctl détecté dans le répertoire.")
        return

    for filename in files:
        file_path = os.path.join(folder, filename)
        
        with open(file_path, 'r', encoding='utf-8') as f:
            lines = f.readlines()
        
        has_changed = False
        new_content = []
        
        for line in lines:
            # .strip() permet de détecter 'dset' même s'il y a des espaces/tabulations
            if line.strip().startswith('dset') or line.strip().startswith('DSET'):
                if '^' in line:
                    # Remplacement du symbole ^ par le chemin
                    line = line.replace('^', absolute_path)
                    has_changed = True
            new_content.append(line)
        
        if has_changed:
            with open(file_path, 'w', encoding='utf-8') as f:
                f.writelines(new_content)
            print(f"✅ Renamed : {filename}")
        else:
            print(f"🟡 Ignored (already done or without '^') : {filename}")

if __name__ == "__main__":
    update_all_ctls(directory, new_path)
```

```python
import os
import requests
from datetime import datetime, timedelta
import os

# --- CONFIGURATION ---
INIT_DAY = 0  # Équivalent à $1
TARGET_DIR = ".."
WEEKS_TO_FETCH = 6

# Liste des URL modèles (le tag {wk} sera remplacé par l'année + semaine)
URL_SOURCES = [
    "ftp://ftp.star.nesdis.noaa.gov/pub/corp/scsb/wguo/data/Blended_VH_4km/VH/VHP.G04.C07.npp.P{wk}.VH.nc",
    "ftp://ftp.star.nesdis.noaa.gov/pub/corp/scsb/wguo/data/VHP_4km/VH/VHP.G04.C07.NP.P{wk}.VH.nc",
    "https://www.star.nesdis.noaa.gov/pub/corp/scsb/wguo/data/Blended_VH_4km/VH/VHP.G04.C07.j01.P{wk}.VH.nc",
    "https://www.star.nesdis.noaa.gov/data/pub0018/VHPdata4users/data/Blended_VH_4km/VH/VHP.G04.C07.j01.P{wk}.VH.nc"
]

os.chdir("C:/Users/HP/Desktop/cmtpy/src/cmt/generate_mask/vhi")
def download_file(url, filename):
    """Télécharge un fichier avec gestion d'erreurs simple."""
    try:
        print(f"Tentative de téléchargement : {url}")
        # Note: Pour le FTP pur, urllib est préférable, mais requests gère très bien le HTTPS
        response = requests.get(url, stream=True, timeout=30)
        if response.status_code == 200:
            with open(filename, 'wb') as f:
                for chunk in response.iter_content(chunk_size=8192):
                    f.write(chunk)
            print(f"Succès : {filename} téléchargé.")
            return True
    except Exception as e:
        print(f"Échec de l'URL : {e}")
    return False

def run_retrieval():
    # Nettoyage des PNG (comme dans le script shell)
    for f in os.listdir(TARGET_DIR):
        if f.endswith(".png"):
            os.remove(f)

    for i in range(1, WEEKS_TO_FETCH + 1):
        # Calcul de la date cible
        nd = INIT_DAY + (i * 7)
        target_date = datetime.now() - timedelta(days=nd)
        
        # Calcul de l'année et de la semaine (Logique GrADS: tranches de 7 jours)
        year = target_date.year
        julian_day = int(target_date.strftime("%j"))
        week_num = min(((julian_day - 1) // 7) + 1, 52)
        
        wk_tag = f"{year}0{week_num:02d}"
        target_filename = f"VHP.G04.C07.j01.P{wk_tag}.VH.nc"

        print(f"\n--- Semaine {i} (Tag: {wk_tag}) ---")

        # Vérification de l'existence locale
        if os.path.exists(target_filename):
            print(f"Le fichier {target_filename} est déjà présent localement.")
            continue

        # Tentative de téléchargement via les différentes sources
        for url_template in URL_SOURCES:
            source_url = url_template.format(wk=wk_tag)
            if download_file(source_url, target_filename):
                break # On s'arrête si un téléchargement réussit

if __name__ == "__main__":
    run_retrieval()
```

```python
import os
import shutil
import pandas as pd
from datetime import datetime, timedelta

# --- 1. CONFIGURATION ---
CNTRY_CODE = "AFR"        # Correspond à $1
CNTRY_NAME = "Africa"    # Correspond à $2
INIT_DAY = 0              # Correspond à $3 (initday)
MASK_OPT = "mask_yes"     # Correspond à $4

# Chemins
VHI_MAPS_DIR = f"vhi_maps"
FIX_DIR = "../fix"

def get_week_logic(days_ago):
    """
    Reproduit la logique de calendrier du script original.
    Calcul des semaines de 7 jours à partir du 1er Janvier.
    """
    target_date = datetime.now() - timedelta(days=days_ago)
    year = target_date.year
    
    # Jour julien (1-366)
    julian_day = int(target_date.strftime("%j"))
    
    # Numéro de semaine (Logique GrADS : tranches de 7 jours)
    week_num = ((julian_day - 1) // 7) + 1
    week_num = min(week_num, 52)
    
    # Calcul des bornes de la semaine pour le titre (du script original)
    # On trouve le jour de la semaine du 1er Janvier de l'année concernée
    jan1 = datetime(year, 1, 1)
    start_day_val = jan1.weekday() + 1 # 1=Lundi, 7=Dimanche
    
    # Calcul du décalage
    cur_wkday = target_date.weekday() + 1
    diff = cur_wkday - start_day_val
    if diff < 0: diff += 7
    
    bgn_date = target_date - timedelta(days=diff)
    end_date = bgn_date + timedelta(days=6)
    
    # Sécurité fin d'année
    if julian_day > 357:
        end_date = datetime(year, 12, 31)
        
    return {
        "year": year,
        "week_str": f"{week_num:02d}",
        "file_tag": f"{year}0{week_num:02d}", # ex: 2026017
        "bgn_str": bgn_date.strftime("%d%b%Y").upper(),
        "end_str": end_date.strftime("%d%b%Y").upper()
    }

def prepare_data_for_week(wk_info, week_idx):
    """Gère les fichiers et prépare les variables pour chaque semaine."""
    #os.chdir("/arc2")
    # 1. Identification du fichier satellite source
    # Le script cherche NPP ou J01
    prefix = "VHP.G04.C07"
    suffix = f"P{wk_info['file_tag']}.VH.nc"
    
    source_file = None
    for sat in ["npp", "j01"]:
        f_name = f"{prefix}.{sat}.{suffix}"
        if os.path.exists(f_name):
            source_file = f_name
            break
    cur = os.getcwd()
    print(f"current path === {cur}")
    if not source_file:
        print(f"!!! Données manquantes : Semaine {wk_info['week_str']} ({wk_info['year']})")
        return

    # 2. Préparation du fichier de travail (vhi.nc)
    shutil.copy(source_file, "vhi.nc")
    
    # 3. Lecture des paramètres de géométrie (équivalent awk)
    ydel = "0.1"
    latlon_file = os.path.join(FIX_DIR, f"{CNTRY_NAME}_latlon")
    if os.path.exists(latlon_file):
        with open(latlon_file, 'r') as f:
            lines = f.readlines()
            if len(lines) >= 2:
                # Colonne 9 (index 8) de la ligne 2
                ydel = lines[1].split()[8]

    # --- LOG ---
    print(f"--- SEMAINE {week_idx} (Sem. {wk_info['week_str']}) ---")
    print(f"Période : {wk_info['bgn_str']} au {wk_info['end_str']}")
    print(f"Source  : {source_file}")
    print(f"YDEL    : {ydel}")    
    #if os.path.exists("vhi.nc"):
    #    os.remove("vhi.nc")
```

```python
import os
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from matplotlib.patches import PathPatch
from shapely.ops import unary_union
import geopandas as gpd
import pandas as pd
import numpy as np
from matplotlib.colors import ListedColormap, BoundaryNorm
from xgrads import open_CtlDataset


def get_vhi_colormap():
    hex_colors = [
        "#FF00FF", # 0-6
        "#E10032", # 6-12
        "#FF7D7D", # 12-24
        "#FFAA00", # 24-36
        "#FFFF64", # 36-48
        "#64FF64", # 48-60
        "#009600", # 60-72
        "#5050FF", # 72-84
        "#0000C8"  # 84-100
    ]
    levels = [0, 6, 12, 24, 36, 48, 60, 72, 84, 100]
    
    cmap = ListedColormap(hex_colors)
    norm = BoundaryNorm(levels, len(hex_colors))
    
    return cmap, norm, levels # On retourne aussi levels pour le plot

def load_config_extent(path_fix):
    print(f"Loading extents")
    dir = os.getcwd()
    print(f"Load extents  CWD: {dir}")
    with open(path_fix, 'r') as f:
        line = f.readline().split()
        country = line[0]
        lat1, lat2 = float(line[1]), float(line[2])
        lon1, lon2 = float(line[3]), float(line[4])
    return {
        "name": country,
        "lat_range": (lat1, lat2),
        "lon_range": (lon1, lon2)
    }


def prepare_vhi_data(path_vhi, path_mask):
    """Charge, divise par 100 et applique le masque pays."""
    
    print(f"Preparing VHI data")
    dir = os.getcwd()
    print(f"VHI CWD: {dir}")
    ds = xr.open_dataset(r'vhi.nc')
    ds = ds.assign_coords({
                "lat": ds.latitude,
                "lon": ds.longitude
            }).rename({
                "HEIGHT": "lat",
                "WIDTH": "lon"
            })
    vhi = ds["VHI"].load()
    
    mask_ds = xr.open_dataset(path_mask)
    land_mask = xr.open_dataset("GLDASp5_landmask_025d.nc4")
    mask_interp = mask_ds['mask_data'].interp(lat=vhi.lat, lon=vhi.lon, method="nearest")
    land_interp = land_mask["GLDAS_mask"].isel(time=-1).interp(lat=vhi.lat, lon=vhi.lon, method="nearest").load()

    return vhi.where((mask_interp == 1) &(land_interp ==1))


def plot_vhi_map(da, extent_info, path_shp, wk_info):
    """Génère et sauvegarde la carte Cartopy."""
    print(f"Generating VHI maps")
    cmap, norm, levels = get_vhi_colormap()
    
    fig = plt.figure(figsize=(11, 8.5))
    ax = plt.axes(projection=ccrs.PlateCarree())
    
    da = da.where(da > 0)
    im = ax.pcolormesh(
        da.lon,
        da.lat, 
        da, 
        #levels=levels,
        cmap=cmap, 
        norm=norm,
        transform=ccrs.PlateCarree(),
        zorder=1)
    
    if os.path.exists(path_shp):
        gdf = gpd.read_file(path_shp)
        gdf.plot(ax=ax, edgecolor='black', facecolor='none', linewidth=0.5, zorder=2)
    else:
        print(f"Attention: {path_shp} introuvable. Utilisation des frontières par défaut.")
        #ax.add_feature(cfeature.BORDERS, linewidth=0.5, edgecolor='black')

    #ax.add_feature(cfeature.COASTLINE, linewidth=0.7)
    #ax.add_feature(cfeature.BORDERS, linestyle='-', alpha=0.8) 
    # Configuration géographique
    ax.set_extent([extent_info['lon_range'][0], extent_info['lon_range'][1], 
                   extent_info['lat_range'][0], extent_info['lat_range'][1]],
                  crs=ccrs.PlateCarree()) # Best practice to specify CRS for extent
                  
    date_val = wk_info['end_str']
    plt.title(f"Vegetation Health Index - {extent_info['name']}",loc='left', fontsize=11, fontweight='bold', pad=10)
    plt.title(f"Week Ending {date_val}", loc='right', fontsize=9, style='italic')

    cbar = fig.colorbar(im, ax=ax, orientation='horizontal', fraction=0.04, pad=0.08, extend='both')
    cbar.set_label('VHI (%)', fontsize=10)
    cbar.ax.set_xticklabels([str(l) for l in levels])
    currentdir = os.getcwd()
    output_path=f"{currentdir}\\vhi_maps\\vhi_wk{wk_info['week_str']}.png"

    plt.savefig(output_path, dpi=200, bbox_inches='tight')
    plt.close(fig)
    print(f"Fichier créé : {output_path}")
```

```python
import os
import requests
import ssl
from urllib3.util.ssl_ import create_urllib3_context
from requests.adapters import HTTPAdapter
from concurrent.futures import ThreadPoolExecutor

# --- CONFIGURATION OPTIONNELLE ---
SOURCES_SELECTIONNEES = ["rfe2"] 

# Classe pour forcer une sécurité SSL plus faible (DH_KEY_TOO_SMALL bypass)
class DESAdapter(HTTPAdapter):
    def init_poolmanager(self, *args, **kwargs):
        context = create_urllib3_context()
        context.load_default_certs()
        # On baisse le niveau de sécurité pour accepter les clés DH anciennes
        context.set_ciphers('DEFAULT@SECLEVEL=1')
        kwargs['ssl_context'] = context
        return super(DESAdapter, self).init_poolmanager(*args, **kwargs)

def download_file(url, target_folder):
    filename = url.split('/')[-1]
    save_path = os.path.join(target_folder, filename)
    os.makedirs(target_folder, exist_ok=True)
    
    # On utilise un session pour appliquer l'Adapter SSL
    session = requests.Session()
    session.mount('https://', DESAdapter())
    
    try:
        # Tentative en HTTPS (avec sécurité réduite)
        response = session.get(url, timeout=20, stream=True)
        
        # Si HTTPS échoue encore, on tente un repli automatique en HTTP
        if response.status_code != 200 and url.startswith("https://sgbd.acmad.org"):
            http_url = url.replace("https://", "http://")
            response = requests.get(http_url, timeout=20, stream=True)

        if response.status_code == 200:
            with open(save_path, 'wb') as f:
                for chunk in response.iter_content(chunk_size=16384):
                    f.write(chunk)
            return f"SUCCÈS: {filename}"
        else:
            return f"ERREUR {response.status_code}: {filename}"
            
    except Exception as e:
        return f"ÉCHEC: {filename} (Erreur: {e})"

def run_spp_retrieval():
    base_urls = [
        "https://sgbd.acmad.org/thredds/fileServer/ACMAD/CDD/climatedataservice/SPP_data/spp_{rndta}/spp_{rndta}_comb_1ic-0proj.ctl",
        "https://sgbd.acmad.org/thredds/fileServer/ACMAD/CDD/climatedataservice/SPP_data/spp_{rndta}/spp_{rndta}_comb_1ic-0proj.dat",
        "https://sgbd.acmad.org/thredds/fileServer/ACMAD/CDD/climatedataservice/SPP_data/spp_{rndta}/spp_{rndta}_comb_1ic-1proj.ctl",
        "https://sgbd.acmad.org/thredds/fileServer/ACMAD/CDD/climatedataservice/SPP_data/spp_{rndta}/spp_{rndta}_comb_1ic-1proj.dat",
        "https://sgbd.acmad.org/thredds/fileServer/ACMAD/CDD/climatedataservice/SPP_data/spp_{rndta}/spp_{rndta}_comb_1ic-2proj.ctl",
        "https://sgbd.acmad.org/thredds/fileServer/ACMAD/CDD/climatedataservice/SPP_data/spp_{rndta}/spp_{rndta}_comb_1ic-2proj.dat",
        "https://sgbd.acmad.org/thredds/fileServer/ACMAD/CDD/climatedataservice/SPP_data/spp_{rndta}/spp_{rndta}_comb_2ic-0proj.ctl",
        "https://sgbd.acmad.org/thredds/fileServer/ACMAD/CDD/climatedataservice/SPP_data/spp_{rndta}/spp_{rndta}_comb_2ic-0proj.dat",
        "https://sgbd.acmad.org/thredds/fileServer/ACMAD/CDD/climatedataservice/SPP_data/spp_{rndta}/spp_{rndta}_comb_2ic-1proj.ctl",
        "https://sgbd.acmad.org/thredds/fileServer/ACMAD/CDD/climatedataservice/SPP_data/spp_{rndta}/spp_{rndta}_comb_2ic-1proj.dat",
        "https://sgbd.acmad.org/thredds/fileServer/ACMAD/CDD/climatedataservice/SPP_data/spp_{rndta}/spp_{rndta}_comb_3ic-0proj.ctl",
        "https://sgbd.acmad.org/thredds/fileServer/ACMAD/CDD/climatedataservice/SPP_data/spp_{rndta}/spp_{rndta}_comb_3ic-0proj.dat",
        "https://sgbd.acmad.org/thredds/fileServer/ACMAD/CDD/climatedataservice/SPP_data/spp_{rndta}/spp_timescales.txt"
    ]

    all_tasks = []
    for rndta in SOURCES_SELECTIONNEES:
        target_dir = f"spp_data_{rndta}"
        for url_template in base_urls:
            all_tasks.append((url_template.format(rndta=rndta), target_dir))

    print(f"Démarrage avec contournement SSL pour : {', '.join(SOURCES_SELECTIONNEES)}")

    with ThreadPoolExecutor(max_workers=3) as executor:
        results = list(executor.map(lambda p: download_file(*p), all_tasks))

    for res in results:
        print(res)

if __name__ == "__main__":
    run_spp_retrieval()
```

```python
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from matplotlib.colors import ListedColormap, BoundaryNorm
import numpy as np

def get_spp_colormaps():
    # Seuils GrADS : 40, 60, 75, 90
    levels = [40, 60, 75, 90, 100]
    
    # Couleurs extraites des codes GrADS du script
    cmaps = {
        'below':  ListedColormap(['#FFFF00', '#FFCC00', '#FF9900', '#FF6600']), # Jaunes/Oranges
        'normal': ListedColormap(['#99FF99', '#33CC33', '#009900', '#006600']), # Verts
        'above':  ListedColormap(['#99CCFF', '#3399FF', '#0066FF', '#000099'])  # Bleus
    }
    norms = {k: BoundaryNorm(levels, v.N) for k, v in cmaps.items()}
    return cmaps, norms, levels

def plot_spp_python(datasets, country_info, captions, output_prefix):
    """
    datasets: liste de 6 xarray.DataArray (p1, p2, p3)
    country_info: dict avec lon1, lon2, lat1, lat2, name
    captions: liste de tuples (titre_a, titre_b)
    """
    cmaps, norms, levels = get_spp_colormaps()
    lon_range = [country_info['lon1'], country_info['lon2']]
    lat_range = [country_info['lat1'], country_info['lat2']]

    for i, ds in enumerate(datasets):
        fig = plt.figure(figsize=(10, 12))
        ax = plt.axes(projection=ccrs.PlateCarree())
        ax.set_extent([*lon_range, *lat_range])

        #--- LOGIQUE DE CALCUL (équivalent GrADS : if(p3 > p1...)) ---
        p1, p2, p3 = ds['p1'].isel(time=-1).load(), ds['p2'].isel(time=-1).load(), ds['p3'].isel(time=-1).load()
        
        # Trouver la probabilité max entre les 3 catégories
        max_prob = np.maximum(np.maximum(p1, p2), p3)
        
        # Masquage : on ne garde que les zones > 33.3% (maskout pp-100/3)
        mask_threshold = 100/3
        
        # Affichage sélectif par catégorie
        # BELOW
        print(f"{p1}")

        break
        p1_masked = p1.where((p1 == max_prob) & (p1 > mask_threshold))
        ax.pcolormesh(ds.lon, ds.lat, p1_masked, cmap=cmaps['below'], norm=norms['below'], transform=ccrs.PlateCarree())
        
        # NORMAL
        p2_masked = p2.where((p2 == max_prob) & (p2 > mask_threshold))
        ax.pcolormesh(ds.lon, ds.lat, p2_masked, cmap=cmaps['normal'], norm=norms['normal'], transform=ccrs.PlateCarree())
        
        # ABOVE
        p3_masked = p3.where((p3 == max_prob) & (p3 > mask_threshold))
        im_above = ax.pcolormesh(ds.lon, ds.lat, p3_masked, cmap=cmaps['above'], norm=norms['above'], transform=ccrs.PlateCarree())

        #--- HABILLAGE (Shapefiles) ---
        ax.add_feature(cfeature.BORDERS, linewidth=1.2, edgecolor='black')
        ax.add_feature(cfeature.COASTLINE, linewidth=0.8)
        # Ajout des subdivisions administratives (adm1/adm2)
        ax.add_feature(cfeature.STATES.with_scale('10m'), linewidth=0.5, edgecolor='gray')

        #--- TITRES ---
        plt.title(f"{captions[i][0]}\n{captions[i][1]}", fontsize=12, fontweight='bold')

        #--- LÉGENDES (Les 3 Colorbars) ---
        # On crée 3 axes au bas de la figure pour les 3 catégories
        cbar_ax1 = fig.add_axes([0.15, 0.08, 0.2, 0.02])
        cbar_ax2 = fig.add_axes([0.40, 0.08, 0.2, 0.02])
        cbar_ax3 = fig.add_axes([0.65, 0.08, 0.2, 0.02])
        
        fig.colorbar(plt.cm.ScalarMappable(norm=norms['below'], cmap=cmaps['below']), cax=cbar_ax1, orientation='horizontal', label='BELOW')
        fig.colorbar(plt.cm.ScalarMappable(norm=norms['normal'], cmap=cmaps['normal']), cax=cbar_ax2, orientation='horizontal', label='NORMAL')
        fig.colorbar(plt.cm.ScalarMappable(norm=norms['above'], cmap=cmaps['above']), cax=cbar_ax3, orientation='horizontal', label='ABOVE')

        # Sauvegarde
        plt.savefig(f"{output_prefix}_{country_info['name']}_{i+1}.png", dpi=200, bbox_inches='tight')
        plt.close()
```

```python
import os
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from xgrads import open_CtlDataset
from matplotlib.colors import ListedColormap, BoundaryNorm

# --- 1. CONFIGURATION & LECTURE DES FICHIERS FIXES ---

def read_country_config(filepath):
    """Équivalent de read(file2) pour récupérer les coordonnées du pays."""
    with open(filepath, 'r') as f:
        data = f.readline().split()
        # On saute l'en-tête GrADS si présent, on prend la ligne de données
        #data = lines[1].split() 
        return {
            'name': data[0],
            'lat1': float(data[1]), 'lat2': float(data[2]),
            'lon1': float(data[3]), 'lon2': float(data[4])
            #'xlint': float(data[5]), 'ylint': float(data[6])
        }

def read_captions(filepath):
    """Lit les titres ligne par ligne (2 lignes par dataset)."""
    captions = []
    with open(filepath, 'r') as f:
        lines = [line.strip() for line in f.readlines() if line.strip()]
        for i in range(0, len(lines), 2):
            captions.append((lines[i], lines[i+1]))
    return captions

# --- 2. DÉFINITION DES COULEURS (Styles GrADS) ---

def get_grads_colors():
    """
    Recrée les échelles de couleurs extraites de l'image :
    - Below: Jaune à Rouge foncé
    - Normal: Vert clair à Vert foncé
    - Above: Bleu clair à Bleu foncé
    """
    # Les seuils correspondent aux étiquettes 40, 60, 75, 90 de votre image
    levels = [33, 40, 60, 75, 90, 100]
    
    cmaps = {
        # BELOW: Jaune (#ffff80), Orange clair (#ffc800), Orange foncé (#ff8000), Rouge (#cc0000)
        'below':  ListedColormap(['#FFFFFF', '#ffff80', '#ffc800', '#ff8000', '#cc0000']),
        
        # NORMAL: Vert pâle (#c8ffc8), Vert clair (#78ff78), Vert moyen (#00cc00), Vert foncé (#007d00)
        'normal': ListedColormap(['#FFFFFF', '#c8ffc8', '#78ff78', '#00cc00', '#007d00']),
        
        # ABOVE: Bleu très clair (#b4ffff), Bleu ciel (#78d2ff), Bleu royal (#0078ff), Bleu marine (#003cff)
        'above':  ListedColormap(['#FFFFFF', '#b4ffff', '#78d2ff', '#0078ff', '#003cff'])
    }
    
    # BoundaryNorm garantit que chaque intervalle [40-60], [60-75], etc., utilise la bonne couleur
    norms = {k: BoundaryNorm(levels, v.N) for k, v in cmaps.items()}
    
    return cmaps, norms

# --- 3. LOGIQUE DE CALCUL ET TRACÉ ---

def run_orchestrator(mask_enabled=True):
    # Fichiers d'entrée
    ctl_files = [
        'spp_cmorph_comb_1ic-0proj.ctl', 'spp_cmorph_comb_1ic-1proj.ctl',
        'spp_cmorph_comb_1ic-2proj.ctl', 'spp_cmorph_comb_2ic-0proj.ctl',
        'spp_cmorph_comb_2ic-1proj.ctl', 'spp_cmorph_comb_3ic-0proj.ctl'
    ]
    
    config = read_country_config('../../Africa_latlon')
    captions = read_captions('spp_timescales.txt')
    cmaps, norms = get_grads_colors()

    # Chargement du masque pays
    mask_ds = None
    if mask_enabled:
        mask_path = '../../gis_resources/country_masks0p1/365dcal/AFR_mask.nc'
        mask_ds = xr.open_dataset(mask_path)

    for i, ctl in enumerate(ctl_files):
        print(f"Processing {ctl}...")
        ds = open_CtlDataset(ctl)
        def lon_360_to_180(ds):
            # On ajuste les longitudes : (lon + 180) % 360 - 180
            ds.coords['lon'] = (ds.coords['lon'] + 180) % 360 - 180.125
            return ds.sortby('lon')

        ds = lon_360_to_180(ds)
        
        # Mapping des variables (on prend le dernier pas de temps comme dans votre GrADS)
        v = list(ds.data_vars)
        p1 = ds[v[0]].isel(time=-1).load()
        p2 = ds[v[1]].isel(time=-1).load()
        p3 = ds[v[2]].isel(time=-1).load()

        # Logique de détermination de la catégorie dominante
        max_p = np.maximum(np.maximum(p1, p2), p3)
        threshold = 100/3.0

        # Application du masque géographique
        if mask_enabled:
            # On s'assure que le masque a les mêmes dimensions que les données
            m = mask_ds['mask_data']
            # Interpole le masque sur la grille du dataset si nécessaire
            if not (m.lat.equals(ds.lat) and m.lon.equals(ds.lon)):
                m = m.interp(lat=ds.lat, lon=ds.lon, method='nearest')
            
            p1, p2, p3 = p1.where(m > 0), p2.where(m > 0), p3.where(m > 0)
            max_p = max_p.where(m > 0)

        # Création de la figure
        fig = plt.figure(figsize=(10, 8))
        ax = plt.axes(projection=ccrs.PlateCarree())
        ax.set_extent([config['lon1'], config['lon2'], config['lat1'], config['lat2']])

        # --- TRACÉ SÉCURISÉ PAR CATÉGORIE ---
        categories = [
            ('below', p1, (p1 == max_p) & (p1 > threshold)),
            ('normal', p2, (p2 == max_p) & (p2 > threshold)),
            ('above', p3, (p3 == max_p) & (p3 > threshold))
        ]

        for cat_name, data, mask in categories:
            plot_data = data.where(mask)
            
            # On ne trace que s'il reste des données (évite Dimensions of C (0,0))
            if not plot_data.isnull().all():
                ax.pcolormesh(
                    ds.lon, ds.lat, plot_data,
                    cmap=cmaps[cat_name], 
                    norm=norms[cat_name],
                    shading='nearest', # SOLUTION à l'erreur de dimensions
                    transform=ccrs.PlateCarree()
                )

        # Habillage
        ax.add_feature(cfeature.COASTLINE)
        ax.add_feature(cfeature.BORDERS, linewidth=1.5)
        #ax.add_feature(cfeature.STATES.with_scale('10m'), linewidth=0.5, edgecolor='gray')

        # Titres
        plt.title(f"{captions[i][0]}\n{captions[i][1]}", fontsize=12, fontweight='bold', pad=15)

        # Colorbars
        for j, cat in enumerate(['below', 'normal', 'above']):
            cax = fig.add_axes([0.12 + j*0.3, 0.08, 0.25, 0.02])
            cb = plt.colorbar(
                plt.cm.ScalarMappable(norm=norms[cat], cmap=cmaps[cat]), 
                cax=cax, orientation='horizontal'
            )
            cb.set_label(cat.upper(), fontsize=10, fontweight='bold')

        # Sauvegarde
        output_name = f"spp_{config['name']}_{i+1}.png"
        plt.savefig(output_name, dpi=200, bbox_inches='tight')
        plt.close()
        print(f"Saved: {output_name}")

if __name__ == "__main__":
    run_orchestrator()
```

```python
import pandas as pd
import io

def clean_station_file(input_path, output_path):
    cleaned_rows = []
    
    with open(input_path, 'r', encoding='utf-8') as f:
        for line in f:
            # Nettoyage des espaces blancs superflus
            parts = line.split()
            
            # On ignore les lignes vides ou trop courtes
            if len(parts) < 4:
                continue
            
            # Structure cible : id_stn, lat, lon, name
            # On prend les 3 premiers éléments (id, lat, lon)
            # Et on prend le 4ème comme nom, en ignorant les répétitions (ex: Rabat Rabat)
            id_stn = parts[0]
            lat = parts[1]
            lon = parts[2]
            name = parts[3] 
            
            cleaned_rows.append(f"{id_stn}\t{lat}\t{lon}\t{name}")

    # Sauvegarde du fichier propre
    with open(output_path, 'w', encoding='utf-8') as f:
        f.write("\n".join(cleaned_rows))
    
    print(f"✅ Fichier nettoyé sauvegardé sous : {output_path}")

# Utilisation
clean_station_file('Africa_stns.txt', 'Africa_stns_clean.txt')
```

```python
import os
import requests

def download_nasa_mask(url):
    # On extrait le nom du fichier à partir de l'URL
    filename = url.split('/')[-1]
    
    print(f"Téléchargement de : {filename}")
    
    try:
        # stream=True permet de ne pas charger tout le fichier en RAM d'un coup
        response = requests.get(url, stream=True, timeout=30)
        response.raise_for_status() # Vérifie si le lien est valide (Erreur 404, etc.)
        
        total_size = int(response.headers.get('content-length', 0))
        block_size = 1024 # 1 Kibibyte
        
        with open(filename, 'wb') as f:
            downloaded = 0
            for data in response.iter_content(block_size):
                downloaded += len(data)
                f.write(data)
                
                # Petit indicateur de progression
                if total_size > 0:
                    done = int(50 * downloaded / total_size)
                    print(f"\r[{'=' * done}{' ' * (50-done)}] {downloaded/1024/1024:.2f} MB", end='')
            
        print(f"\n\nSuccès ! Fichier enregistré sous : {os.path.abspath(filename)}")
        return True

    except Exception as e:
        print(f"\nErreur lors du téléchargement : {e}")
        return False

# Lancement
url_mask = "https://ldas.gsfc.nasa.gov/sites/default/files/ldas/gldas/VEG/GLDASp5_landmask_10d.nc4"
download_nasa_mask(url_mask)
```

```python
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

def visualize_land_mask(file_path):
    # 1. Ouverture du fichier
    # On utilise 'decode_times=False' car les masques n'ont souvent pas de dimension temporelle standard
    ds = xr.open_dataset(file_path)
    
    # Affichage de la structure pour identifier le nom de la variable (souvent 'GLDAS_mask')
    print("Structure du fichier :")
    print(ds)
    
    # On récupère la variable du masque (à ajuster selon le print(ds) ci-dessus)
    # Dans ce fichier NASA, la variable s'appelle généralement 'landmask' ou 'GLDAS_mask'
    var_name = list(ds.data_vars)[0]
    mask = ds[var_name]

    # 2. Création de la figure
    fig = plt.figure(figsize=(12, 6))
    ax = plt.axes(projection=ccrs.PlateCarree())
    
    # 3. Affichage des données
    # On utilise 'nearest' pour un masque binaire (0 ou 1) afin d'éviter le flou
    im = mask.plot(
        ax=ax, 
        transform=ccrs.PlateCarree(), 
        cmap='Greens', 
        add_colorbar=True,
        cbar_kwargs={'label': 'Land Mask (1=Land, 0=Water)', 'shrink': 0.7}
    )

    # 4. Habillage géographique
    ax.add_feature(cfeature.COASTLINE, linewidth=0.8)
    ax.add_feature(cfeature.BORDERS, linestyle=':', alpha=0.5)
    ax.gridlines(draw_labels=True, dms=True, x_inline=False, y_inline=False, alpha=0.3)

    plt.title(f"Visualisation du masque NASA : {var_name}", fontweight='bold')
    plt.show()

# Utilisation
path = "GLDASp5_landmask_025d.nc4"
visualize_land_mask(path)
```

```python


def plot_precip(rsl, rsl_name, country_iso, country, rndta):

    # =========================================================
    # 1. PATHS
    # =========================================================
    base_dir = Path.cwd().parent / "data"
    mask_path = base_dir / "gis_resources" / f"country_masks{rsl_name}" / "365dcal" / f"{country_iso}_mask.nc"
    if rndta.lower() == "arc2":
        daily_precip_path = (base_dir / "ARC2" / "arc2" / "arc2.ctl")#.resolve()
        clim_path = (base_dir / "ARC2" / "arc2_clim" / "arc2_clim.ctl")#.resolve()
        precip_var = "pmer2"
    elif rndta.lower() == "rfe2": 
        daily_precip_path = (base_dir / "rfe2_data" / "rfe2_daily" / "rfe2daily.ctl")#.resolve()
        clim_path = (base_dir / "rfe2_data" / "rfe2_clim" / "rfe2clim.ctl")#.resolve()
        precip_var = "r"
    elif rndta.lower() == "cmorph": 
        daily_precip_path = (base_dir / "CMORPH" / "cmorph_daily" / "cmorph.ctl")#.resolve()
        clim_path = (base_dir / "CMORPH" / "cmorph_clim" / "cmorphclim.ctl")#.resolve()
        precip_var = "r"
    shap_path = base_dir / "gis_resources" / "countries" / f"{country_iso}_adm" / f"{country_iso}_adm1.shp"
    mask_path =fr"../data/gis_resources/country_masks{rsl_name}/365dcal/{country_iso}_mask.nc"
    # =========================================================
    # 2. DATA LOAD & REPARATION UNIQUE SHP (Hors boucle)
    # =========================================================
    print("Loading data...")
    print(f"mask loading")
    mask_nc = xr.open_dataset(mask_path)
    print(f"mask loaded")
    #daily_data = open_CtlDataset(os.path.relpath(daily_precip_path))
    #clim_data = open_CtlDataset(os.path.relpath(clim_path))

    abs_daily_path = daily_precip_path.resolve()
    abs_clim_path = clim_path.resolve()

    # 4. Normalisation pour contrer le bug de xgrads sous Windows
    if platform.system() == "Windows":
        # Force l'usage de slashes '/' pour empêcher xgrads d'ajouter './' devant 'C:/'
        final_daily_path = abs_daily_path.as_posix()
        final_clim_path = abs_clim_path.as_posix()
    else:
        # Format natif pour Linux et macOS (contient déjà des slashes '/')
        final_daily_path = str(abs_daily_path)
        final_clim_path = str(abs_clim_path)
    final_daily_path = r"../data/CMORPH/cmorph_daily/cmorph.ctl"
    final_clim_path = r"../data/CMORPH/cmorph_clim/cmorphclim.ctl"
    # 5. Ouverture sécurisée des fichiers de données GrADS
    daily_data = open_CtlDataset(final_daily_path)
    clim_data = open_CtlDataset(final_clim_path)
    #print(f"daily data : {daily_data}")
    print("*********************")
    print(f"clim data : {clim_data}")


    # Chargement et réparation unique du Shapefile
    shp_adm1 = gpd.read_file(shap_path)
    shp_adm1 = shp_adm1.to_crs("EPSG:4326")
    shp_adm1["geometry"] = shp_adm1.geometry.apply(make_valid)
    shp_adm1 = shp_adm1[shp_adm1.is_valid]

    bounds = shp_adm1.total_bounds
    
    # Fusion des contours pour obtenir la frontière nationale externe
    country_geom = shp_adm1.geometry.unary_union
    if not country_geom.is_valid:
        country_geom = country_geom.buffer(0)

    # Création du masque inversé (le calque de cache extérieur)
    world_box = sgeom.box(bounds[0]-10, bounds[1]-10, bounds[2]+10, bounds[3]+10)
    inverse_mask = world_box.difference(country_geom)

    periodes = [7, 10, 30, 60, 90, 180]

    # =========================================================
    # 3. COLORS & LIMITS
    # =========================================================
    colors_precip = [
        "#FFFFFF", "#D3FFCB", "#93F78F", "#4BC34B",
        "#C3F3FB", "#73B7F7", "#4B8BEF", "#E3E3FF",
        "#B3A3FF", "#8D80E3", "#FFFBBB", "#FFB333",
        "#E74333", "#B73333", "#EBA3A3", "#FFEBEB"
    ]
    limits_precip = [2, 5, 10, 25, 50, 75, 100, 150, 200, 300, 500, 750, 1000, 1500, 2500]

    anom_colors = [
        "#C30000", "#FF3300", "#FFA200", "#FFEB7A",
        "#7A5147", "#B78E84", "#F4DFD5", "#FFFFFF",
        "#CBFFC1", "#7AF975", "#1EB71E", "#98D5FE",
        "#2984F4", "#DFDFFF", "#8272EF"
    ]
    limits_anom = [-500, -300, -200, -100, -50, -25, -10, 10, 25, 50, 100, 200, 300, 500]
    # 1. Graduation exacte de l'image (niveaux non linéaires)
    percent_levels = [1, 5, 25, 50, 80, 120, 150, 200, 400, 600]

    # 2. Codes HEX précis extraits directement de l'image
    percent_colors = [
        "#E1BEB4",  # 0 à 1    : Rose / Beige clair
        "#C00000",  # 1 à 5    : Rouge vif
        "#FF3200",  # 5 à 25   : Orange rouge
        "#FFA000",  # 25 à 50  : Orange classique
        "#FFE878",  # 50 à 80  : Jaune pastel
        "#FFFFFF",  # 80 à 120 : Blanc (Neutre)
        "#C8FFBE",  # 120 à 150: Vert très clair
        "#78F573",  # 150 à 200: Vert pomme
        "#1EB41E",  # 200 à 400: Vert franc
        "#96D2FA",  # 400 à 600: Bleu ciel
        "#2882F0",  # 600 à 1000: Bleu foncé / Roi
    ]

    mask_sorted = mask_nc['mask_data']

    pad = 0.5
    extent_box = [bounds[0]-pad, bounds[2]+pad, bounds[1]-pad, bounds[3]+pad]

    out_dir = base_dir / "spatial_maps" / country / rndta 
    out_dir.mkdir(parents=True, exist_ok=True)

    # =========================================================
    # 4. LOOP PERIODS
    # =========================================================
    for p in periodes:
        print(f"\n Period : {p} days")

        if rndta =="cmorph":
            daily_slice = daily_data[precip_var].isel(time=slice(-p, None), lev =-1)
            clim_slice = clim_data[precip_var].isel(time=slice(-p, None), lev =-1)
        else:
            daily_slice = daily_data[precip_var].isel(time=slice(-p, None))
            clim_slice = clim_data[precip_var].isel(time=slice(-p, None))

        pcur = p * daily_slice.mean(dim='time').load()
        pclim = p * clim_slice.mean(dim='time').load()
        
        time_values = daily_slice.time
        start_date = pd.to_datetime(time_values[1].values).strftime('%d%b%Y')
        end_date = pd.to_datetime(time_values[-1].values).strftime('%d%b%Y')

        anom = pcur - pclim
        pclimrate = pclim 
        drymask = xr.where(pclimrate >= 0.25, 1, 0)

        pcnp = ((pcur / pclim.where(pclim > 1)) * 100)
        pcnp = pcnp.where(np.isfinite(pcnp)).clip(max=400)
        pcnp_clean = pcnp.where(drymask == 1)

        mask_resized = mask_sorted.interp(
            lat=pcur.lat,
            lon=pcur.lon,
            method="nearest"
        ).fillna(0)
        
        # Application des fonctions de lissage
        anom = apply_clean_smoothing(anom, sigma=0)
        pcnp = apply_clean_smoothing(pcnp_clean, sigma=0)
        pcur = apply_clean_smoothing(pcur, sigma=0)
        pclim = apply_clean_smoothing(pclim, sigma=0)
        
        anom_masked = anom.where(drymask == 1) #& (mask_resized == 1))
        pcur_masked = pcur.where(drymask == 1) #& (mask_resized == 1))
        pclim_masked = pclim.where(drymask == 1)# & (mask_resized == 1))
        pcnp_masked = pcnp_clean.where(mask_resized == 1)

        plot_configs = {
            "precip_anomaly": {"data": anom_masked, "levels": limits_anom, "colors": anom_colors, "title": "Anomaly", "unit": "mm"},
            "total_precip": {"data": pcur_masked, "levels": limits_precip, "colors": colors_precip, "title": "Current Precip", "unit": "mm"},
            "normal_precip": {"data": pclim_masked, "levels": limits_precip, "colors": colors_precip, "title": "Climatology", "unit": "mm"},
            "percent_normal_precip": {"data": pcnp_masked, "levels": percent_levels, "colors": percent_colors, "title": "Percent of Normal", "unit": "%"}
        }

        # =========================================================
        # 5. DESSIN MULTI-COUCHES PERFORMANCE
        # =========================================================
        for key, config in plot_configs.items():
            print(f"{config['title']}")

            fig, ax = plt.subplots(figsize=(12, 9.27), subplot_kw={'projection': ccrs.PlateCarree()})

            ax.set_extent(extent_box, crs=ccrs.PlateCarree())
            
            # 🟡 COULEUR EXTERNE INITIALE
            #ax.set_facecolor("#e0e0e0")
            ax.set_facecolor("#ffffff")

            # =====================================================================
            # 🟢 ÉTAPE 0 : LE TAPIS DE FOND INTERNE (Zones Sèches Masquées)
            # =====================================================================
            # Dessine la forme du pays complète en blanc (zorder 1) sous les données.
            # Les zones exclues par le drymask (NaN) hériteront de ce blanc impeccable.
            ax.add_geometries([country_geom], ccrs.PlateCarree(),
                              facecolor="#e0e0e06c", edgecolor='none', zorder=1)
            # =====================================================================

            # Étape B : Dessin de la matrice de données cartographiques (zorder 2)
            im = ax.contourf(
                config['data'].lon,
                config['data'].lat,
                config['data'],
                levels=config['levels'],
                colors=config['colors'],
                extend='both',
                transform=ccrs.PlateCarree(),
                zorder=2
            )

            # =====================================================================
            # 🟡 APPLICATION DU MASQUE INVERSÉ EXTERNE (zorder 5)
            # =====================================================================
            # Recouvre hermétiquement toutes les coulures de lissage hors frontières
            ax.add_geometries([inverse_mask], ccrs.PlateCarree(), 
                              facecolor="#ffffff", edgecolor='none', zorder=5)

            # =====================================================================
            # 🔴 TRACÉ CHIRURGICAL ET STRATIFIÉ DES FRONTIÈRES (zorder 6 & 7)
            # =====================================================================
            # 1. Frontières administratives régionales internes (Discrètes)
            ax.add_geometries(shp_adm1.geometry, ccrs.PlateCarree(),
                              facecolor='none', edgecolor='#4a4a4a', linewidth=0.5, zorder=6)
            
            # 2. Frontière nationale principale (Contour souverain net au premier plan)
            ax.add_geometries([country_geom], ccrs.PlateCarree(),
                              facecolor='none', edgecolor='black', linewidth=1.5, zorder=7)

            # Barre d'échelle
            cbar = plt.colorbar(im, ax=ax, fraction=0.03, pad=0.04)
            cbar.set_label(config['unit'])

            # Grille de coordonnées
            gl = ax.gridlines(draw_labels=True, alpha=0.6)
            gl.top_labels = False
            gl.right_labels = False

            # Titre appliqué (Rouge de style GrADS)
            if key == "normal_precip":
                plt.title(f"{rndta.upper()} - {config['title']} : {p}-day \n Period: {start_date[:-4]} - {end_date[:-4]}", color='red')
            else:
                plt.title(f"{rndta.upper()} - {config['title']} : {p}-day \n Period: {start_date} - {end_date}", color='black')

            # Sauvegarde et libération immédiate de la RAM
            plt.savefig(out_dir / f"{country}_{p}day_{rndta}_{key}.png", dpi=150, bbox_inches='tight')
            plt.clf()
            plt.close(fig)

            print("✅ Saved")
            
    print(f"✅{rndta} Precipitation maps generated.")
```

```python
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg') # Pour ne pas afficher de fenêtre
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.ticker import MultipleLocator

def process_data(data):
    """Calcule les données journalières et les cumuls."""
    s = pd.Series(data.values)
    s = s.replace(-999.0, np.nan).clip(lower=0)
    daily = s.fillna(0)
    cumul = s.fillna(0).cumsum()
    return daily, cumul

def get_nice_step(max_val, is_cumul=False):
    """Calcule l'échelle intelligente des axes Y."""
    imax = int(round(max_val + (max_val / 50.0)))
    target_del = imax / 10.0
    steps = [1, 2, 3, 4, 5, 10, 15, 20, 25, 30, 40, 50, 60, 70, 80, 90, 100]
    if is_cumul:
        steps += [125, 175, 250, 350, 500]
    for s in steps:
        if s >= target_del: return imax, s
    return imax, steps[-1]

def generate_rainfall_plot(pcur_ts, pclim_ts, stnnm, lt, ln, stncnt, period):
    """Génère et sauvegarde le graphique final."""
    r1, r2 = process_data(pcur_ts)
    r3, r4 = process_data(pclim_ts)
    
    dates = pcur_ts.time.values
    r4_high, r4_low = 1.2 * r4, 0.8 * r4
    
    imax1, del1 = get_nice_step(max(r1.max(), r3.max()))
    imax2, del2 = get_nice_step(max(r2.max(), r4.max()), is_cumul=True)
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 11), gridspec_kw={'height_ratios': [2, 1]})
    plt.subplots_adjust(hspace=0.20)

    #print(f"############ R2 Vs R4_high START #############")
    #print(f"R2: {r2}")
    #print(f"R4_high : {r4_high}")
    #print(f"############ R2 Vs R4_high END #############")

    # --- Panneau Supérieur ---
    ax1.fill_between(dates, r2, r4_high, where=(r2 > r4_high), interpolate=True, color='#32CD32', alpha=0.5, label='120% Of Normal')
    ax1.fill_between(dates, r2, r4_low, where=(r2 < r4_low), interpolate=True, color='#A52A2A', alpha=0.5, label='80% Of Normal')
    ax1.fill_between(dates, r4_low, r4_high, color="#8B8B95", alpha=0.3)

    perc = (r2.iloc[-1] / r4.iloc[-1]) * 100 if r4.iloc[-1] != 0 else 0
    perf_txt = f"Current Performance: {perc:.1f}% of Normal"

    ax1.plot(dates, r4, color='gray', linestyle='--', linewidth=2, label='Climatological Normal')
    ax1.plot(dates, r2, color='black', linewidth=1, label=perf_txt)
    
    # --- Panneau Inférieur ---
    ax2.bar(dates, r1, color='#4682B4', width=0.7)

    # Formattage des axes
    my_fmt = mdates.DateFormatter('%d %b')
    for i, ax in enumerate([ax1, ax2]):
        ax.tick_params(labelbottom=True)
        ax.xaxis.set_major_formatter(my_fmt)
        ax.grid(True, linestyle='-', alpha=0.5)
        ax.legend(loc='upper left', fontsize='small')
        plt.setp(ax.get_xticklabels(), rotation=30, ha='right')
        ax.set_xlim(dates[0], dates[-1])
        
        if i == 0: # AX1
            ax.set_ylim(0, imax2 * 1.05)
            ax.yaxis.set_major_locator(MultipleLocator(del2))
            ax.set_ylabel('Rainfall vs Normal (mm)', fontweight='bold')
        else: # AX2
            ax.set_ylim(0, imax1)
            ax.yaxis.set_major_locator(MultipleLocator(del1))
            ax.set_ylabel('Rainfall (mm)', fontweight='bold')

    plt.suptitle(f"ARC2 Point Time Series\n{len(dates)}-Day Rainfall @ {stnnm} ({lt}N, {ln}E)", fontsize=14, y=0.94)
    
    filename = f"{stncnt}_{period}.png"
    plt.savefig(filename, dpi=100, bbox_inches='tight')
    plt.close(fig)
    return filename
```

```python
def calculate_pixels(lat, lon, lat1, lat2, lon1, lon2):
    # 1. Simuler 'q w2xy' : On normalise la position entre 0 et 1 
    # (On suppose ici que la page GrADS couvre exactement l'emprise lat/lon)
    margin_x = 0.8
    margin_y = 0.6
    # x_inch : position horizontale (0 à 11 pouces)
    x_inch = ((lon - lon1) / (lon2 - lon1)) * (11.0 + 2*margin_x)
    
    # y_inch : position verticale (0 à 8.5 pouces)
    y_inch = ((lat - lat1) / (lat2 - lat1)) * (8.5 - 2*margin_y)
    
    # 2. Appliquer tes formules spécifiques :
    # xxx1 = (800 / 11) * xx * 1.5
    xxxx1 = (800 / 11.0) * x_inch * 1.5
    
    # yyy1 = (618 / 8.5) * (8.5 - yy) * 1.5
    yyyy1 = (618 / 8.5) * (8.5 - y_inch) * 1.5
    
    return round(xxxx1, 2), round(yyyy1, 2)
    #return xxxx1, yyyy1
```

```python
def w2xy(lon, lat, lons, lats):
    # lons and lats are 1D arrays defining the grid
    dx = lons[1] - lons[0]
    dy = lats[1] - lats[0]

    x = (lon - lons[0]) / dx
    y = (lat - lats[0]) / dy

    return x, y
```

```python
def calculate_pixels_final(lat, lon, lat1, lat2, lon1, lon2):
    # 1. Marges standard GrADS sur une page 11x8.5 (en pouces)
    # Si 'set parea' n'est pas utilisé, GrADS utilise environ :
    parea_left = 1.0   # Marge pour les labels de latitude
    parea_right = 10.5 # Marge à droite
    parea_bottom = 0.75 # Marge pour les labels de longitude
    parea_top = 7.75    # Marge en haut
    
    # 2. Position dans la zone de dessin (en pouces)
    xx = parea_left + ((lon - lon1) / (lon2 - lon1)) * (parea_right - parea_left)
    yy = parea_bottom + ((lat - lat1) / (lat2 - lat1)) * (parea_top - parea_bottom)
    
    # 3. Conversion en pixels (formule GrADS printim)
    # x = (Largeur / 11) * pouces * facteur_echelle
    xxxx1 = (800 / 11.0) * xx * 1.5
    # y = (Hauteur / 8.5) * (Hauteur_Page - pouces) * facteur_echelle
    yyyy1 = (618 / 8.5) * (8.5 - yy) * 1.5
    
    return xxxx1, yyyy1
```

```python
def calculate_pixels(lat, lon, lat1, lat2, lon1, lon2):
    # --- LES MARGES PAR DÉFAUT DE GRADS (Crucial) ---
    # Ces valeurs simulent le cadre interne de GrADS sur une page de 11x8.5
    parea_left = 1.1   # Marge à gauche (pour les labels de latitude)
    parea_right = 10.5 # Marge à droite
    parea_bottom = 0.75 # Marge en bas (pour les labels de longitude)
    parea_top = 7.75    # Marge en haut
    
    # 1. Conversion Lat/Lon vers Pouces (Simulation de 'q w2xy')
    # On calcule la position relative du point dans le cadre de GrADS
    xx = parea_left + (lon - lon1) / (lon2 - lon1) * (parea_right - parea_left)
    yy = parea_bottom + (lat - lat1) / (lat2 - lat1) * (parea_top - parea_bottom)
    
    # 2. Conversion Pouces vers Pixels (La formule de votre script)
    # xxx1 = (800 / 11) * xx * 1.5
    xxxx1 = (800 / 11.0) * xx * 1.5
    
    # yyyy1 = (618 / 8.5) * (8.5 - yy) * 1.5
    # (On inverse l'axe Y car en image le 0 est en haut)
    yyyy1 = (618 / 8.5) * (8.5 - yy) * 1.5
    
    return xxxx1, yyyy1
```

```python
def calculate_pixels_equidistant(lat, lon, lat1, lat2, lon1, lon2):
    # Dimensions de l'image finale
    img_width = 1200
    img_height = 927
    
    # Marges réelles (estimées pour GrADS par défaut)
    margin_x = 120  # Pixels réservés à gauche/droite
    margin_y = 80   # Pixels réservés en haut/bas
    
    # Largeur et hauteur utiles de la carte en pixels
    map_width = img_width - (2 * margin_x)
    map_height = img_height - (2 * margin_y)
    
    # Calcul de la position
    # (lon - lon1) / (lon2 - lon1) donne un ratio entre 0 et 1
    x_pix = margin_x + ((lon - lon1) / (lon2 - lon1)) * map_width
    
    # Pour l'axe Y, on inverse (lat2 en haut, lat1 en bas)
    y_pix = margin_y + ((lat2 - lat) / (lat2 - lat1)) * map_height
    
    return x_pix, y_pix
```

```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import geopandas as gpd
import pandas as pd
from xgrads import open_CtlDataset
import os
from scipy.ndimage import gaussian_filter
import time

# --- 1. FONCTIONS OPTIMISÉES ---
start_time = time.time()

def apply_clean_smoothing(da, sigma=1.0):
    """Lissage gaussien préservant le masque."""
    if sigma <= 0: return da
    mask = da.notnull()
    # Utilisation directe des valeurs numpy pour la vitesse
    smoothed = gaussian_filter(da.fillna(0).values, sigma=sigma)
    return xr.DataArray(smoothed, coords=da.coords, dims=da.dims).where(mask)

def lon_360_to_180(ds):
    """Ajustement rapide des longitudes de 0-360 vers -180-180."""
    return ds.assign_coords(lon=(((ds.lon + 180) % 360) - 180)).sortby('lon')

# --- 2. CONFIGURATION ET CHARGEMENT ---

PATH_SPI = r"..\spi\data\spi_cmorph"
PATH_MASK_NC = r"C:\Users\HP\Desktop\cmtpy\ClimToolKit\gis_resources\countries\AFR_adm\country_masks0p25\365dcal\AFR_cmask.nc"
PATH_SHP = r"C:\Users\HP\Desktop\cmtpy\ClimToolKit\gis_resources\countries\AFR_adm\AFR_adm1.shp"

# Chargement unique des ressources statiques
mask_ds = xr.open_dataset(PATH_MASK_NC)['mask_data'].sortby('lat')
shp_africa = gpd.read_file(PATH_SHP)
bounds = shp_africa.total_bounds

spi_levels = [-2, -1.6, -1.3, -0.8, -0.5, 0.5, 0.8, 1.3, 1.6, 2]
spi_colors = ['#B20000', '#FF0000', '#FF6600', '#FFBD33', '#FFFF99', 
              '#FFFFFF', '#B2EBF2', '#80B3FF', '#3385FF', '#1A53FF', '#311B92']

# --- 3. BOUCLE DE TRAITEMENT ---

periods = [1, 3, 6, 12, 24]

for p in periods:
    print(f">>> Processing: SPI {p}-Month")
    
    # Ouverture des fichiers
    ds_spi = lon_360_to_180(open_CtlDataset(os.path.join(PATH_SPI, f'cmorph.spi.{p}.mo.ctl')))
    ds_m = lon_360_to_180(open_CtlDataset(os.path.join(PATH_SPI, f'drymask{p}.ctl')))

    # Extraction du dernier pas de temps
    spi = ds_spi['p'].isel(time=-1, lev=0).load()
    
    # Correction cruciale : Alignement du mois pour le drymask
    current_month = pd.to_datetime(spi.time.values).month
    # Si le drymask a 12 mois (clim), on prend le mois correspondant, sinon le dernier
    if len(ds_m.time) == 12:
        dm = ds_m['dm'].isel(time=current_month-1, lev=0).load()
    else:
        dm = ds_m['dm'].isel(time=-1, lev=0).load()

    # Nettoyage drymask (-999 -> NaN)
    dm = dm.where(dm != -999.)

    # --- 4. ALIGNEMENT DES GRILLES (Optimisé) ---
    # On aligne tout sur la grille du SPI pour éviter de créer des pixels inutiles
    mask_resized = mask_ds.interp(lat=spi.lat, lon=spi.lon, method="nearest").fillna(0)
    dm_resized = dm.interp(lat=spi.lat, lon=spi.lon, method="nearest")

    # Lissage
    spi_smooth = apply_clean_smoothing(spi, sigma=.2)

    # Masquage final
    # On garde les données là où (Pays == 1) ET (Pas un désert permanent != 1)
    spi_masked = spi_smooth.where((mask_resized == 1) & (dm_resized >= -1.))

    # --- 5. PLOT ---
    fig = plt.figure(figsize=(12, 9))
    ax = plt.axes(projection=ccrs.PlateCarree())
    
    # Fond de carte optionnel (paysages)
    # ax.add_feature(cfeature.COASTLINE, linewidth=0.5)
    
    im = ax.contourf(
        spi_masked.lon, spi_masked.lat, spi_masked,
        levels=spi_levels,
        colors=spi_colors,
        extend='both',
        transform=ccrs.PlateCarree()
    )

    # Ajustement automatique de la vue sur le SHP
    ax.set_extent([bounds[0], bounds[2], bounds[1], bounds[3]], crs=ccrs.PlateCarree())
    shp_africa.plot(ax=ax, edgecolor='black', facecolor='none', linewidth=0.5, zorder=10)

    # Cosmétique
    gl = ax.gridlines(draw_labels=True, alpha=0.2, linestyle='--')
    gl.top_labels = gl.right_labels = False

    date_str = pd.to_datetime(spi.time.values).strftime("%b %Y")
    plt.title(f"CMORPH ADJ 00Z SPI\n{p}-Month Period Ending {date_str}", fontsize=14, pad=20)
    
    # Colorbar optimisée
    cbar = fig.colorbar(im, ax=ax, orientation='vertical', fraction=0.02, pad=0.04)
    cbar.set_label('Standardized Index')

    # Sauvegarde
    out_name = f"SPI_{p}mo_{date_str.replace(' ', '_')}.png"
    plt.savefig(out_name, dpi=150, bbox_inches='tight')
    plt.close()
    print(f"Done: {out_name}")
duration = time.time() - start_time

print(f"--- All periods completed in :{duration} seconds ")
```

```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import geopandas as gpd
import pandas as pd
from xgrads import open_CtlDataset
import os
from scipy.ndimage import gaussian_filter
from concurrent.futures import ProcessPoolExecutor
import time

# --- CONFIGURATION DES CHEMINS ---
PATH_SPI = r"..\spi\data\spi_cmorph"
PATH_MASK_NC = r"C:\Users\HP\Desktop\cmtpy\ClimToolKit\gis_resources\countries\AFR_adm\country_masks0p25\365dcal\AFR_cmask.nc"
PATH_SHP = r"C:\Users\HP\Desktop\cmtpy\ClimToolKit\gis_resources\countries\AFR_adm\AFR_adm1.shp"

start_time = time.time()
# --- FONCTIONS OUTILS ---
def apply_clean_smoothing(da, sigma=1.2):
    if sigma <= 0: return da
    mask = da.notnull()
    # On passe en float32 pour la rapidité
    smoothed = gaussian_filter(da.fillna(0).astype(np.float32).values, sigma=sigma)
    return xr.DataArray(smoothed, coords=da.coords, dims=da.dims).where(mask)

def lon_360_to_180(ds):
    return ds.assign_coords(lon=(((ds.lon + 180) % 360) - 180)).sortby('lon')

# --- FONCTION PRINCIPALE PAR PÉRIODE ---
def process_one_period(p):
    """Cette fonction contient tout le processus pour UNE période donnée."""
    try:
        print(f"Démarrage SPI {p}-Month...")
        
        # 1. Chargement des données
        ds_spi = lon_360_to_180(open_CtlDataset(os.path.join(PATH_SPI, f'cmorph.spi.{p}.mo.ctl')))
        ds_m = lon_360_to_180(open_CtlDataset(os.path.join(PATH_SPI, f'drymask{p}.ctl')))
        
        # Chargement du masque pays (chargé à l'intérieur pour éviter les conflits de threads)
        mask_ds = xr.open_dataset(PATH_MASK_NC)['mask_data'].sortby('lat')
        
        # 2. Extraction du dernier pas de temps
        spi = ds_spi['p'].isel(time=-1, lev=0).load()
        current_month = pd.to_datetime(spi.time.values).month
        
        # Sélection intelligente du mois dans le drymask
        if len(ds_m.time) == 12:
            dm = ds_m['dm'].isel(time=current_month-1, lev=0).load()
        else:
            dm = ds_m['dm'].isel(time=-1, lev=0).load()
        
        # Nettoyage
        dm = dm.where(dm != -999.)
        
        # 3. Alignement des grilles (On réduit le masque vers la grille SPI)
        mask_resized = mask_ds.interp(lat=spi.lat, lon=spi.lon, method="nearest").fillna(0)
        dm_resized = dm.interp(lat=spi.lat, lon=spi.lon, method="nearest")

        # 4. Lissage et Masquage
        spi_smooth = apply_clean_smoothing(spi, sigma=1.2)
        # Condition : Pays == 1 ET Pas un désert (!= 1) pour voir les négatifs et positifs
        spi_masked = spi_smooth.where((mask_resized == 1) & (dm_resized != 1))

        # 5. Dessin
        spi_levels = [-2, -1.6, -1.3, -0.8, -0.5, 0.5, 0.8, 1.3, 1.6, 2]
        spi_colors = ['#B20000', '#FF0000', '#FF6600', '#FFBD33', '#FFFF99', 
                      '#FFFFFF', '#B2EBF2', '#80B3FF', '#3385FF', '#1A53FF', '#311B92']

        fig = plt.figure(figsize=(10, 8))
        ax = plt.axes(projection=ccrs.PlateCarree())
        
        im = ax.contourf(
            spi_masked.lon, spi_masked.lat, spi_masked,
            levels=spi_levels, colors=spi_colors, extend='both'
        )

        # Ajout SHP
        shp_africa = gpd.read_file(PATH_SHP)
        bounds = shp_africa.total_bounds
        ax.set_extent([bounds[0], bounds[2], bounds[1], bounds[3]], crs=ccrs.PlateCarree())
        shp_africa.plot(ax=ax, edgecolor='black', facecolor='none', linewidth=0.5, zorder=10)

        # Cosmétique
        date_str = pd.to_datetime(spi.time.values).strftime("%b %Y")
        plt.title(f"CMORPH SPI {p}-Month\nEnding {date_str}", fontsize=12)
        plt.colorbar(im, ax=ax, fraction=0.03, pad=0.04)

        # Sauvegarde
        out_name = f"SPI_{p}mo_{date_str.replace(' ', '_')}.png"
        plt.savefig(out_name, dpi=120, bbox_inches='tight')
        plt.close()
        
        return f"Succès : {out_name}"
    
    except Exception as e:
        return f"Erreur sur SPI {p}-Month : {str(e)}"

# --- BLOC D'EXÉCUTION PARALLÈLE ---
if __name__ == "__main__":
    periods = [1, 3, 6, 12, 24]
    
    print(f"Lancement du traitement parallèle sur {os.cpu_count()} cœurs potentiels...")
    
    # ProcessPoolExecutor lance chaque période dans un processus séparé
    with ProcessPoolExecutor(max_workers=len(periods)) as executor:
        results = list(executor.map(process_one_period, periods))
    
    for res in results:
        print(res)
    duration = time.time() - start_time
    print(f"--- Toutes les tâches sont terminées  en : {duration}---")
```

```python
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import geopandas as gpd
import pandas as pd
import numpy as np
from matplotlib.colors import ListedColormap, BoundaryNorm

# --- 1. CONFIGURATION DES DONNÉES ---
# Chemins (à adapter selon votre structure)
PATH_VHI = "vhi.nc"  # GrADS 'open vhi.ctl'
#PATH_MASK = "../gis_resources/country_masks0p036/senegal_vhimasks0p036.nc"
chemin =r"C:\Users\HP\Desktop\cmtpy\ClimToolKit\gis_resources\country_masks0p036\365dcal\SEN_cmask.nc"
PATH_MASK = chemin
# Lecture du masque en .nc (NetCDF)
#mask_nc = xr.open_dataset(chemin)

PATH_SHP = "../gis_resources/countries/SEN_adm/SEN_adm1.shp"
PATH_FIX = "../fix/senegal_latlon"

# --- 2. LECTURE DES PARAMÈTRES (Équivalent du fichier latlon) ---
# On utilise pandas pour lire les coordonnées de la zone d'étude
fix_data = pd.read_csv(PATH_FIX, sep=r'\s+', header=None, skiprows=1)
country = fix_data.iloc[0, 0]
lat1, lat2 = fix_data.iloc[0, 1], fix_data.iloc[0, 2]
lon1, lon2 = fix_data.iloc[0, 3], fix_data.iloc[0, 4]

# --- 3. CHARGEMENT ET TRAITEMENT ---
ds_vhi = xr.open_dataset(PATH_VHI)
vhi_raw = ds_vhi['VHI'].isel(time=0) # 'set t 1'

# Dans GrADS 'd vhi/100', les données satellites NOAA sont souvent stockées en entiers
vhi = vhi_raw / 100.0

# Chargement du masque (senegal_vhimasks0p036)
ds_mask = xr.open_dataset(PATH_MASK)
# On aligne le masque sur la grille VHI
mask = ds_mask['mask_data'].interp(lat=vhi.lat, lon=vhi.lon, method="nearest")

# Application du masque (masksplot = 1)
vhi_masked = vhi.where(mask == 1)

# --- 4. CONFIGURATION DES COULEURS (Équivalent 'set rgb') ---
vhi_colors = [
    (255/255, 0/255, 200/255),   # 151
    (236/255, 19/255, 79/255),   # 152
    (248/255, 160/255, 160/255), # 153
    (255/255, 201/255, 14/255),  # 154
    (255/255, 255/255, 128/255), # 155
    (128/255, 255/255, 128/255), # 156
    (0/255, 128/255, 0/255),     # 157
    (66/255, 66/255, 255/255),   # 158
    (0/255, 0/255, 180/255)      # 159
]
vhi_levels = [6, 12, 24, 36, 48, 60, 72, 84] # 'set clevs'
cmap = ListedColormap(vhi_colors)
norm = BoundaryNorm([0] + vhi_levels + [100], cmap.N)

# --- 5. CARTOGRAPHIE ---
fig = plt.figure(figsize=(10, 8))
ax = plt.axes(projection=ccrs.PlateCarree())

# Tracé Shaded (Grfill)
im = ax.pcolormesh(
    vhi_masked.lon, vhi_masked.lat, vhi_masked,
    cmap=cmap, norm=norm, transform=ccrs.PlateCarree()
)

# Habillage
ax.set_extent([lon1, lon2, lat1, lat2], crs=ccrs.PlateCarree())
ax.add_feature(ccrs.feature.COASTLINE, linewidth=0.8)
ax.add_feature(ccrs.feature.BORDERS, linestyle=':')

# Ajout du SHP (SEN_adm1)
shp = gpd.read_file(PATH_SHP)
shp.plot(ax=ax, edgecolor='black', facecolor='none', linewidth=1.5)

# Titres
date_dt = pd.to_datetime(vhi_raw.time.values)
date_str = date_dt.strftime("%d %b %Y")
plt.title("Vegetation Health Index", fontsize=14, fontweight='bold', pad=20)
plt.title(f"For The Week Ending {date_str}", loc='right', fontsize=10)

# Colorbar (xcbar)
cbar = fig.colorbar(im, ax=ax, orientation='horizontal', fraction=0.05, pad=0.07, extend='both')
cbar.set_label('VHI Index (%)')

# Sauvegarde
plt.savefig('vhi.png', dpi=300, bbox_inches='tight')
plt.show()
```
