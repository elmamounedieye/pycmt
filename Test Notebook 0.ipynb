# This is to delete existing version of pycmt

````python
import os
import platform
import shutil
from pathlib import Path

print("Uninstalling existing package via pip...")
os.system("pip uninstall pycmt pycmt -y")

home = Path.home()
system_os = platform.system().lower()
paths_to_clean = []

# Ciblage rigoureux des répertoires 'data' internes selon l'OS et l'environnement
if system_os == "windows":
    appdata_local = Path(os.environ.get("LOCALAPPDATA", home / "AppData" / "Local"))
    appdata_roaming = Path(os.environ.get("APPDATA", home / "AppData" / "Roaming"))
    
    paths_to_clean.extend([
        appdata_roaming / "Python" / "Python313" / "site-packages" / "pycmt" / "data",
        appdata_local / "Programs" / "Python" / "Python312" / "Lib" / "site-packages" / "pycmt" / "data",
        
        appdata_roaming / "Python" / "Python313" / "site-packages" / "pycmt",
        appdata_roaming / "Python" / "Python313" / "site-packages" / "pycmt_cpc-0.1.0.dist-info",
        appdata_local / "Programs" / "Python" / "Python312" / "Lib" / "site-packages" / "pycmt",
        appdata_local / "Programs" / "Python" / "Python312" / "Lib" / "site-packages" / "pycmt_cpc-0.1.0.dist-info",
        appdata_local / "Programs" / "Python" / "Python312" / "share" / "jupyter" / "labextensions" / "@jupyter-widgets" / "jupyterlab-manager",
        appdata_local / "pip" / "cache"
    ])

elif system_os == "linux":
    paths_to_clean.extend([
        home / ".local" / "lib" / "python3.10" / "site-packages" / "pycmt" / "data",
        home / ".local" / "lib" / "python3.12" / "site-packages" / "pycmt" / "data",
        home / "anaconda3" / "envs" / "cpc_stcnn_env" / "lib" / "python3.10" / "site-packages" / "pycmt" / "data",
        home / "anaconda3" / "lib" / "python3.12" / "site-packages" / "pycmt" / "data",
        
        home / ".local" / "lib" / "python3.10" / "site-packages" / "pycmt",
        home / ".local" / "lib" / "python3.12" / "site-packages" / "pycmt",
        home / "anaconda3" / "envs" / "cpc_stcnn_env" / "lib" / "python3.10" / "site-packages" / "pycmt",
        home / "anaconda3" / "lib" / "python3.12" / "site-packages" / "pycmt",
        home / ".cache" / "pip"
    ])

elif system_os == "darwin":  # macOS
    paths_to_clean.extend([
        Path("/opt/homebrew/Caskroom/miniforge/base/lib/python3.13/site-packages/pycmt/data"),
        home / "Library" / "Python" / "3.10" / "lib" / "python" / "site-packages" / "pycmt" / "data",
        home / "Library" / "Python" / "3.12" / "lib" / "python" / "site-packages" / "pycmt" / "data",
        home / "miniconda3" / "envs" / "cpc_stcnn_env" / "lib" / "python3.10" / "site-packages" / "pycmt" / "data",
        
        Path("/opt/homebrew/Caskroom/miniforge/base/lib/python3.13/site-packages/pycmt"),
        home / "Library" / "Python" / "3.10" / "lib" / "python" / "site-packages" / "pycmt",
        home / "Library" / "Python" / "3.12" / "lib" / "python" / "site-packages" / "pycmt",
        home / "miniconda3" / "envs" / "cpc_stcnn_env" / "lib" / "python3.10" / "site-packages" / "pycmt",
        home / "Library" / "Caches" / "pip"
    ])

# Détection dynamique des dossiers de métadonnées d'installation (.dist-info)
base_search_paths = [
    home / ".local/lib", 
    home / "anaconda3", 
    home / "miniconda3", 
    home / "Library/Python",
    Path("/opt/homebrew/Caskroom/miniforge")
]

for base_path in base_search_paths:
    if base_path.exists():
        for p in base_path.glob("**/pycmt_cpc-*.dist-info"):
            paths_to_clean.append(p)

# Boucle d'exécution du nettoyage du répertoire data et des structures
for path in paths_to_clean:
    if path.exists():
        try:
            if path.is_file():
                os.remove(path)
            else:
                shutil.rmtree(path)
            print(f"✅ Deleted : {path}")
        except Exception as e:
            print(f"⚠️ Impossible to delete {path} : {e}")

print("\nCleaning done!.")
````

````python
#### This function downloads and plots SPI
generate_spis(country, ISO)
````
