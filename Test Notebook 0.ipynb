````python
country = "Country name" #E.g. Senegal, Niger, Cameroon
rndta = "arc2" #arc2 or rfe2 : string
sp_rsl = [0.036, 0.037, 0.25, 0.5, 0.1, 1.0] ### Spatial resolution: select a value according to the country size from the given range
ts_rsl = [0.3, 0.5, 0.7, 0.1] # Time series spatial resolution: increase to 10 the resolution if Africa, select a value
````

````python
######## This function donwloads rainfall data and computes the spatial precip distribution and time series
######## As relative output, we return the country iso code that will be used
ISO = plot_precip_ts(country, rndta, sp_rsl, ts_rsl)
````

````python
#### This function downloads and plots SPP
generate_spp(country, ISO)
````

````python
#### This function downloads and plots SPI
generate_spi_(country, ISO)
````

````python
#### This function downloads and plots VHI
generated_vhi(country, ISO)
````

````python
#### Generating Dashboard
generate_dashboard(country, rndta)
````
